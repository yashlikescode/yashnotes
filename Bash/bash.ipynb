## awk
basic syntax: ```awk 'pattern { action }' file```

````python
# $0 for entire line, $1 for first field, $2 for second field, etc.

!awk '{print $0}' file.txt
````

````python
# Condition / pattern matching

!awk '$2 > 25 {print $0}' file.txt
````

````python
# 3 Main Parts of awk

'''

BEGIN {
    # runs once before processing
}

pattern {
    # runs for matching records
}

END {
    # runs once after processing
}

'''
````

````python

!awk 'BEGIN {print "---Names---"} {print $1} END {print "---end---"}' file.txt
````

````python
# $NF = number of fields, also represents last filed and can be printed

!awk '{print $NF}' file.txt
````

````python
# Second last field can be printed using $(NF-1)

!awk '{print $(NF-1)}' file.txt
````

````python
!awk 'NF != 3 {print "Bad line:", $0}' file.txt
````

````python
# NR = (serial) number of records, can be printed

!awk '{print NR, "-", $0}' file.txt
````

````python
# While using multiple files, FNR = (serial) number of records in current file, can be printed
# FNR resets to 1 when a new file is started, while NR continues to increment

!awk 'BEGIN{print "FNR", "NR", "Line"}{print FNR, NR, $0}' file.txt file2.txt
````

````python
# FS = field separator, default is space, can be changed to any character

!awk 'BEGIN{FS=","} {print $1, $2}' file.csv
````

````python
# we can also use -F option to set field separator

!awk -F',' '{print $1, $3}' file.csv
````

````python
#  OFS = output field separator, default is space, can be changed to any character

!awk -F',' 'BEGIN {OFS=" | "} {print $1, $3}' file.csv
````

````python
# FS  = Field Separator
# OFS = Output Field Separator

# RS  = Record Separator (Nromally newline, can be changed to any character)
# ORS = Output Record Separator (Normally newline, can be changed to any character)
````

````python
# Filtering

!awk '$2 > 25 {print $1}' file.txt
````

````python
# Conditions

'''

==
!=
>
<
>=
<=

'''
````

````python
# AND / OR conditions

'''
&&
||
!
'''

!awk '$2 > 25 && $2 < 35 {print $1}' file.txt
````

````python
# Pattern matching
# Use ~ for matching and !~ for not matching

!awk '$3 ~ /ERROR/ {print $0}' application.log
````

````python
!awk '$3 !~ /ERROR/ {print $0}' application.log
````

````python
# Regex can also be used for pattern matching

!awk '/Worker/ {print $0}' application.log
````

````python
!awk '/session|memory usage/ {print $0}' application.log
````

````python
# For searching a column

!awk '$3 ~ /ERROR/ {print $0}' application.log
````

````python
# User printf for formatted output

# common format specifiers:
# %s   string
# %d   integer
# %f   floating point
# %.2f two decimal places

!awk '{printf "%-10s %5d\n", $1, $2}' file.txt

# -10 (in %-10s): Formats field $1 as a string with a minimum width of 10 characters, left-aligned.
# 5 (in %5d): Formats field $2 as a decimal integer with a minimum width of 5 characters, right-aligned.
````

````python
# Aritehmetic operations can be performed on fields

!awk '{sum += $2} END {print "Total:", sum}' file.txt
````

````python
# same as 'wc -l file.txt' - counts the number of lines in a file

!awk 'END {print NR}' file.txt
````

````python
# Count matching lines

!awk '$3 ~ /ERROR/ {count++} END {print "Matching lines:", count}' application.log
````

````python
# count[$2]++ tracks how many times each unique value in column 2 appears on matching lines.
# count is a map / key-value pair, where the key is the value in column 2 and the value is the count of occurrences.

!awk '/ERROR/ {count[$2]++} END {for (x in count) print x, count[x]}' error.log
````

````python
!awk '{ if ($2 > 80) print $1, "HIGH"; else print $1, "NORMAL" }' file.txt
````

### next
```next``` skips the rest of the current record.

````python
!awk '{if ($3 == "[ERROR]") next; else print $0}' application.log
````

````python
# exit

!awk '{if ($3 == "[ERROR]") exit; else print $0}' application.log
````

````python
# FILENAME can be used to print the name of the current file being processed. FNR can be used to print the line number of the current file being processed.

!awk '{print FILENAME, FNR, $0}' file.txt file2.txt
````

````python
# command line variables can be passed to awk using -v option
# it is better than passing hard coded values in the script, as it makes the script more flexible and reusable.

!awk -v threshold=80 '$2 > threshold {print $0}' file.txt
````

````python
# shell variables can also be passed

!threshold=80;awk -v threshold="$threshold" '$2 > threshold {print $0}' file.txt
````

````python
# sub(string, regex, replacement): Replaces the first occurrence of regex in string with replacement.

!awk '{sub("%","",$1); print $1}' file.txt
````

### awk cheatsheet

````python
# First field
# awk '{print $1}' file

# Last field
# awk '{print $NF}' file

# Number of fields
# awk '{print NF}' file

# Line number
# awk '{print NR, $0}' file

# Filter
# awk '$2 > 100 {print}' file

# Regex
# awk '/ERROR/' file

# Regex in field
# awk '$3 ~ /ERROR/ {print}' file

# CSV
# awk -F',' '{print $1, $3}' file.csv

# Custom output separator
# awk -F',' 'BEGIN {OFS=" | "} {print $1,$3}' file.csv

# Sum
# awk '{sum += $1} END {print sum}' file

# Average
# awk '{sum += $1} END {print sum/NR}' file

# Count
# awk '{count++} END {print count}' file

# Count matching
# awk '/ERROR/ {count++} END {print count}' file

# Frequency
# awk '{count[$1]++} END {for (x in count) print x,count[x]}' file

# Skip header
# awk 'NR > 1 {print}' file

# Multiple files
# awk '{print FILENAME, FNR, $0}' *.log

# Environment/parameter
# awk -v threshold=500 '$2 > threshold {print}' file

# Skip record
# awk '$1 == "DEBUG" {next} {print}' file

# Stop
# awk '/ERROR/ {print; exit}' file

# Replace first
# awk '{sub("foo","bar"); print}' file

# Replace all
# awk '{gsub("foo","bar"); print}' file
````

## grep
Take input → test every line → print lines that match a pattern  
syntax: ```grep [OPTIONS] PATTERN [FILE...]```

````python
!grep "ERROR" application.log
````

````python
# case insensitive search

!grep -i "error" application.log
````

````python
# Invert match
# Show lines that do not match using -v option

!grep -i -v "error" application.log
````

````python
# Show line numbers using -n option

!grep -i -v -n "error" application.log
````

````python
# Count matching lines using -c option

!grep -c "ERROR" application.log
````

````python
# Show matching text using -o option

!grep -o "ERROR" application.log
````

````python
# Regex can also be used for pattern matching

!grep -oE '[0-9]+' application.log
````

````python
# Match whole words using -w option

!grep -w "ERROR" application.log
````

````python
# Match whole lines using -x option

!grep -x "ERROR" application.log
````

````python
# multiple patterns can be searched using -e option

!grep -e "ERROR" -e "WARN" application.log
````

````python
# Search for fixed strings using -F option

!grep -F "I/O" application.log
````

````python
# For regex, use -E option for extended regex, or -P option for Perl regex.

!grep -F "." application.log
````

````python
# Historically, egrep meant extended grep.
````

````python
# Basic regular expressions (BRE) are the default in grep. Extended regular expressions (ERE) can be used with the -E option, and Perl-compatible regular expressions (PCRE) can be used with the -P option.

# Literal text: "hello" - matches exactly "hello" in the input text.
# Wildcard: "." - matches any single character. "h.llo" matches "hello", "hallo", "hxllo", etc.
# Beginning of line: "^" - matches the start of a line. "^hello" matches "hello" only if it appears at the beginning of a line.
# End of line: "$" - matches the end of a line. "hello$" matches "hello" only if it appears at the end of a line.
# Character class: "[abc]" - matches any single character that is either "a", "b", or "c".
# Range: "[a-z]" - matches any single character in the range "a" to "z".
# Negation: "[^abc]" - matches any single character that is not "a", "b", or "c".
# Quantifiers: "*", "+", "?", "{n}", "{n,}", "{n,m}" - specify how many times the preceding element can occur.
# * - matches zero or more occurrences of the preceding element. "ab*" matches "a", "ab", "abb", "abbb", etc.
# + - matches one or more occurrences of the preceding element. "ab+" matches "ab", "abb", "abbb", etc., but not "a".
# ? - matches zero or one occurrence of the preceding element. "ab?" matches "a" or "ab", but not "abb".
# {n} - matches exactly n occurrences of the preceding element. "a{3}" matches "aaa" but not "aa" or "aaaa".
# {n,m} - matches between n and m occurrences of the preceding element. "a{2,4}" matches "aa", "aaa", or "aaaa" but not "a" or "aaaaa".

# OR - "|" - matches either the expression before or after the "|". "cat|dog" matches either "cat" or "dog".
# Grouping - "(...)" - groups expressions together. "(ab|cd)e" matches "abe" or "cde".


# Common regex shortcuts:
# [[:alnum:]] - matches any alphanumeric character (letters and digits).
# [[:alpha:]] - matches any alphabetic character (letters).
# [[:digit:]] - matches any digit (0-9).
# [[:lower:]] - matches any lowercase letter (a-z).
# [[:upper:]] - matches any uppercase letter (A-Z).
# [[:space:]] - matches any whitespace character (spaces, tabs, newlines).
````

````python
# use -r to search recursively in directories.

!grep -r "ERROR" /var/log
````

````python
# search filenames instead of contents using -l option
!grep -l "ERROR" /var/log/*
````

````python
# search filenames that do not contain the pattern using -L option
!grep -L "ERROR" /var/log/*
````

````python
# Include / Exclude files using --include and --exclude options

!grep -r \
  --include="*.java" \
  --include="*.xml" \
  --exclude="*.class" \
  "database" ../../../. | head -n 4
````

````python
# Context around matches
# -B t to show t lines before the match
# -A t to show t lines after the match
# -C t to show t lines before and after the match

!grep -C 2 "Exception" application.log
````

````python
# Treat binary as text using -a option

!grep -a "ERROR" application.log
````

````python
#  -I tells grep to ignore binary files. It is equivalent to --binary-files=without-match.

!grep -I "hello" application.log
````

````python
# Supresses error messages using -s option

!grep -rs "hello" file
````

````python
# Quiet mode using -q option, suppresses all output, only return exit status

!grep -q "hello" file && echo "Pattern found" || echo "Pattern not found"
````

````python
# grep Exit Codes
# 0 → match found
# 1 → no match
# 2 → error

!grep "ERROR" application.log; echo $?
````

## sed
```sed``` = Stream EDitor  
It reads input line by line, applies an operation, and writes the result.

````python
# s means substitute
# s/pattern/replacement/flags

!sed 's/ERROR/ERRORS/' application.log
````

````python
# add additional g to replace all occurrences of the pattern in each line

!sed 's/ERROR/ERRORS/g' application.log
````

````python
# Replace only the second occurence

!sed 's/ERROR/ERRORS/2' application.log
````

````python
# Case-insensitive replacement using I flag

!sed 's/ERROR/ERRORS/I' application.log
````

````python
# the delimiter / is not mandatory, any character can be used as a delimiter, for example, | or #. This is useful when the pattern or replacement contains /.
# sed 's|foo|bar|' file
# sed 's#foo#bar#' file
````

````python
# d removes the line from the output, but does not modify the input file. To modify the input file, use -i option.

!sed '/ERROR/d' application.log
````

````python
# to delete 5th line from the output
# sed '5d' file

# to delete multiple lines, use comma to specify range of lines, for example, to delete lines 5 to 10, use '5,10d'
# sed '5,10d' file

# to delete last line 
# sed '$d' file
````

````python
# use p to print

!sed '5p' file.txt
````

````python
# use comma to print a range

!sed -n '2,3p' file.txt
````

````python
# use $ to print last line

!sed -n '$p' file.txt

# equivalent to - tail -1 file.txt
````

````python
# lines matching to a pattern

!sed -n '/ERROR/p' application.log
````

## find
find searches for files and directories recursively inside a directory tree based on conditions such as:
-filename, type, size, permissions, owner/group, modification time, access time, timestamps, depth, regular expressions, and more
- find [starting-path] [conditions] [actions]

````python
# Find everything under current directory and subdirectories

!find .
````

````python
# -name is a case sensitive search

!find . -name "*.log"
````

````python
# use -iname for case insensitive search

!find . -iname "*.LOG"
````

````python
# Use -type to specify what kind of filesystem object you want.
````

| Option | Meaning          |
| ------ | ---------------- |
| `f`    | regular file     |
| `d`    | directory        |
| `l`    | symbolic link    |
| `b`    | block device     |
| `c`    | character device |
| `p`    | named pipe       |
| `s`    | socket           |


````python
# files only
!find . -type f
````

````python
# directories only
!find . -type d
````

````python
# use -path to search based on path

!find . -path "./*" -type f
````

````python
# use -ipath to make it case-insensitive

!find . -ipath "*/logs/*"
````

````python
# use -maxdepth to declare the depth of search to perform
# 1 -> only current directory

!find . -maxdepth 1 -type f
````

````python
# -mindepth is used to prevent recursing on or before that

# -mindepth 1 -> do not recurse . itself

# We can have precise controle over recursion
!find . -mindepth 1 -maxdepth 2 -type f
````

````python
# search by size

# Larger than 100MB
!find /var/log -type f -size +100M

# Smaller than 10mb
!find /var/log -type f -size -10M

# Exactly approx 10 mb
!find . -type f -size 10M
````

| Unit | Meaning |
| ---- | ------- |
| `c`  | bytes   |
| `k`  | KiB     |
| `M`  | MiB     |
| `G`  | GiB     |


````python
# -empty flag is used for listing only empth directories

!find . -empty
````

````python
# -mtime is used to find files based on modified time

!find . -type f -mtime -1

# -1     less than 1 unit (1 unit = 24 hours period) 
# +7     greater than 7 units
# 7      approximately that range
````

````python
# -mmin is is used based on above units, but the unit here is minute rather than days

# Modified within last 10 minutes
!find . -type f -mmin -10

# Not modified for 60 minutes
!find . -type f -mmin +60
````

````python
# Searching file based on access time

# Find files accessed recently:
!find . -type f -atime -1

# Minute version:
!find . -type f -amin -30
````

````python
# use -ctime to search based on change time

# For example, changes to:
# - permissions
# - ownership
# - file metadata
# - link count
# - sometimes content-related metadata

!find . -type f -ctime -1
````
