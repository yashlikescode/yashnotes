# Data Structures


## Lists

```python
arr = []
print(arr)
```

```python
arr.append(2)
print(arr)
```

```python
arr.extend([3, 4, 5])
print(arr)
```

```python
arr.insert(1, 10)
print(arr)
```

```python
arr.remove(3)
print(arr)
```

```python
temp = arr.pop()
print(temp)
print(arr)
```

```python
print(arr.index(10))
```

```python
print(arr.count(2))
```

```python
arr.sort()
print(arr)
```

```python
arr.reverse()
print(arr)
```

```python
brr = arr.copy()
print(brr)
print(arr)
```

```python
arr.clear()
print(arr)
```

### List Comprehensions

```python
arr = [1, 2, 3, 4, 5]
```

```python
squares = [x * x for x in arr]
print(squares)
```

```python

even = [x for x in arr if x % 2 == 0]
print(even)
```
