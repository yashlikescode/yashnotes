## Sets

````python
# declaring set
s = {1, 2, 3, 4, 5}

# empty set
s1 = set()
````

````python
# adding elements to set
s.add(6)

print(s)  # Output: {1, 2, 3, 4, 5, 6}
````

````python
# update set with multiple elements
s.update([7, 8])

print(s)  # Output: {1, 2, 3, 4, 5, 6, 7, 8}
````

````python
s.remove(3)

print(s)  # Output: {1, 2, 4, 5, 6, 7, 8}
````

````python
s.discard(10)  # Does nothing, no error

print(s)  # Output: {1, 2, 4, 5, 6, 7, 8}
````

````python
a = s.pop()  # Removes and returns an arbitrary element
print(f"a: {a}")  # Output: a: 1 (or any other element)
````

````python
union_s = s.union({9, 10})

print(union_s)  # Output: {1, 2, 4, 5, 6, 7, 8, 9, 10}
````

````python
intersection_s = s.intersection({4, 5, 6, 11})

print(intersection_s)  # Output: {4, 5, 6}
````

````python
# A.difference(B) = A-B
difference_s = s.difference({4, 5, 6, 11})

print(difference_s)  # Output: {1, 2, 7, 8}
````

````python
# A.symmetric_difference(B) = (A-B) U (B-A) | elements that are in either A or B but not in both
symmetric_diff_s = s.symmetric_difference({4, 5, 6, 11})

print(symmetric_diff_s)  # Output: {1, 2, 7, 8, 11}
````

````python
print(s.issuperset({1, 2}))  
# Check if s is a superset of {1, 2}
````

````python
print(s.issubset({1, 2, 4, 5, 6, 7, 8}))  
# Check if s is a subset of {1, 2, 4, 5, 6, 7, 8}
````

````python
s.clear()  # Removes all elements from the set

print(s)  # Output: set()
````

## Dictionaries

````python
d = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(d['name'])  # Output: Alice
print(d)
````

````python
print(d.get('age'))  # Output: 30
````

````python
print(d.keys())  # Output: dict_keys(['name', 'age', 'city'])
````

````python
print(d.values())  # Output: dict_values(['Alice', 30, 'New York'])
````

````python
print(d.items())  # Output: dict_items([('name', 'Alice'), ('age', 30), ('city', 'New York')])
````

````python
d.update({'age': 31, 'country': 'USA'})
print(d)  # Output: {'name': 'Alice', 'age': 31, 'city': 'New York', 'country': 'USA'}
````

````python
age = d.pop('age')  # Remove the 'age' key and return its value
print(age)  # Output: 30
print(d)  # Output: {'name': 'Alice', 'city': 'New York', 'country': 'USA'}
````

````python
last_item = d.popitem()  # Removes and returns the last inserted key-value pair
print(last_item)  # Output: ('country', 'USA')
print(d)  # Output: {'name': 'Alice', 'age': 31, 'city': 'New York'}
````

````python
d.setdefault('country', 'USA')  # If 'country' doesn't exist, set it to 'USA'
print(d)  # Output: {'name': 'Alice', 'age': 31, 'city': 'New York', 'country': 'USA'}
````

````python
newd = d.copy()  # Create a shallow copy of the dictionary
print(newd)  # Output: {'name': 'Alice', 'age': 31, 'city': 'New York', 'country': 'USA'}
````

````python
d.clear()  # Removes all items from the dictionary
print(d)  # Output: {}
````

````python
# Frequency calculation using dictionary

text = "hello world hello"

frequency = {}

for word in text.split():
    frequency[word] = frequency.get(word, 0) + 1
    
print(frequency)  # Output: {'hello': 2, 'world': 1}
````

## Queue / Dequeue

````python
from collections import deque

queue = deque()

print(queue)  # Output: deque([])
````

````python
queue.append(1)
queue.append(2)
queue.append(3)

print(queue)  # Output: deque([1, 2, 3])
````

````python
right_popped = queue.pop()
left_popped = queue.popleft()

print(f"Right popped: {right_popped}")  # Output: Right popped: 3
print(f"Left popped: {left_popped}")  # Output: Left popped: 1
print(queue)  # Output: deque([2])
````

## Functions

````python
def is_palindrome(s):
    # Remove spaces and convert to lowercase
    s = s.replace(" ", "").lower()
    
    # Check if the string is equal to its reverse
    return s == s[::-1]
````

````python
# default argument example
def greet(name = "World"):
    return f"Hello, {name}!"
````

````python
# variable arguements example
def sum_numbers(*args, **kwargs):
    return sum(args)


# kwargs captures {'precision': 2, 'debug': True}, but sum_numbers ignores it
print(sum_numbers(1.234, 5.678, precision=2, debug=True))  # Returns 6.912
````

## Scope -- LEGB
python resolves names using:  
```
L = Local  
E = Enclosing  
G = Global  
B = Built-in
```

````python
x = 10

def outer():
    x = 20

    def inner():
        print(x)

    inner()

outer()

# inner() finds x in the enclosing scope.
# outer x can't be printed directly because it's not in the local scope of inner().
````

````python
# global keyword allows you to modify a variable outside of the current scope.

x = 10

def change():
    global x
    
    print(f"Before change: x = {x}")
    x = 20
    print(f"After change: x = {x}")

change()
````

````python
# nonlocal keyword allows you to modify a variable in the nearest enclosing scope that is not global.

def counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment()

print(counter())
````
